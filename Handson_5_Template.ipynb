## 7.2 Wizard Inventory
Create a program that keeps track of the items that a wizard can carry.

### Console:
```powershell
The Wizard Inventroy Program

COMMAND MENU
walk - Walk down the path 
show - Show all items 
drop - Drop an item
exit - Exit program

Command: walk
While walking down a path, you see a scroll of uncursing. 
Do you want to grab it? (y/n): y
You picked up a scroll of uncursing.

Command: walk
While walking down a path, you see an unknown potion. 
Do you want to grab it? (y/n): y
You can't carry any more items. Drop something first.

Command: show
1.	a wooden staff
2.	a scroll of invisibility
3.	a crossbow
4.	a scroll of uncursing

Command: drop Number: 3
You dropped a crossbow.

Command: exit 
Bye!
```

### Specifications:
- Use the attached text file named `wizard_all_items.txt` that contains a list of all the items that a wizard can carry. (in the corresponding directory)
- When the user selects the walk command, the program should read the items from the file, randomly pick one, and give the user the option to grab it.
- Your program should create another file that stores the items that the wizard is carrying (`wizard_inventory.txt`). Make sure to update this file every time the user grabs or drops an item.
- The wizard can only carry **four** items at a time.
- For the drop command, display an error message if the user enters an invalid number for the item.



````python
import random

with open("wizard_all_items.txt", "w") as f:
    f.write("a wooden staff\na wizard hat\na crossbow\nan unknown potion\n")
with open("wizard_inventory.txt", "w") as f:
    f.write("a wooden staff\n")

def get_all_items():
    file = open("wizard_all_items.txt", "r")
    items = file.read().splitlines()
    file.close()
    return items

def get_inventory():
    file = open("wizard_inventory.txt", "r")
    inventory = file.read().splitlines()
    file.close()
    return inventory

def write_inventory(inventory):
    file = open("wizard_inventory.txt", "w")
    for item in inventory:
        file.write(item + "\n")
    file.close()

def show(inventory):
    number = 1
    for item in inventory:
        print(number, ". ", item)
        number = number + 1

def walk(inventory, all_items):
    item = random.choice(all_items)
    print("While walking down a path, you see " + item + ".")
    choice = input("Do you want to grab it? (y/n): ").lower()

    if choice == "y":
        if len(inventory) >= 4:
            print("You can't carry any more items. Drop something first.")
        else:
            inventory.append(item)
            write_inventory(inventory)
            print("You picked up " + item)

def drop(inventory):
    userInput = input("Number: ")

    if userInput.isdigit():
        index = int(userInput)

        if index < 1 or index > len(inventory):
            print("Invalid item number.")
        else:
            removed = inventory.pop(index - 1)
            write_inventory(inventory)
            print("You dropped " + removed)
    else:
        print("Please enter a valid number")

def main():
    all_items = get_all_items()
    inventory = get_inventory()

    print("The Wizard Inventory Program")
    print("\nCOMMAND MENU")
    print("walk - Walk down the path")
    print("show - Show all items")
    print("drop - Drop an item")
    print("exit - Exit proram")

    while True:
        print()
        command = input("Command: ").lower()

        if command == "walk":
            walk(inventory, all_items)
        elif command == "show":
            show(inventory)
        elif command == "drop":
            drop(inventory)
        elif command == "exit":
            print("Bye!")
            break
        else:
            print("Invalid command. Please try again.")

if __name__ == "__main__":
    main()
````

## 7.3 Monthly Sales
Create a program that reads the sales for 12 months from a file and calculates the total yearly sales as well as the average monthly sales. In addition, this program should let the user edit the sales for any month.

### Console:

### Specifications:
- Use the attached text file named CSV file named `monthly_sales.csv` that contains the month and sales data shown above.
- For the edit command, display an error message if the user doesn’t enter a valid three-letter abbreviation for the month.
- When the user edits the sales amount for a month, the data should be saved to the CSV file immediately. That way, no data is lost, even if the program crashes later.
- Round the results of the monthly average to a maximum of 2 decimal digits.

````python
import csv
import os

def read_sales():
    sales = []
    if os.path.exists('monthly_sales.csv'):
        file = open('monthly_sales.csv', 'r', newline='')
        reader = csv.reader(file)
        for row in reader:
            sales.append(row)
        file.close()
    else:
        print("monthly_sales.csv not found.")
    return sales

def write_sales(sales):
    file = open('monthly_sales.csv', 'w', newline='')
    writer = csv.writer(file)
    writer.writerows(sales)
    file.close()

def show_monthly(sales):
    for row in sales:
        print(row[0] + " - " + row[1])

def show_yearly(sales):
    total = 0.0
    for row in sales:
        total = total + float(row[1])
    
    average = total / len(sales)
    print("Yearly total:    " + str(total))
    print("Monthly average: " + str(round(average, 2)))

def edit_sales(sales):
    month_input = input("Three-letter Month: ")
    month_input = month_input.capitalize()
    
    found = False
    for row in sales:
        if row[0] == month_input:
            new_amount = input("Sales Amount: ")
            row[1] = new_amount
            write_sales(sales)
            print("Sales amount for " + month_input + " was modified.")
            found = True
            break
            
    if found == False:
        print("Invalid three-letter abbreviation.")

def main():
    print("Monthly Sales program")
    sales = read_sales()

    if len(sales) == 0:
        return

    print("\nCOMMAND MENU")
    print("monthly - View monthly sales")
    print("yearly  - View yearly summary")
    print("edit    - Edit sales for a month")
    print("exit    - Exit program")

    while True:
        print()
        command = input("Command: ").lower()
        
        if command == "monthly":
            show_monthly(sales)
        elif command == "yearly":
            show_yearly(sales)
        elif command == "edit":
            edit_sales(sales)
        elif command == "exit":
            print("Bye!")
            break
        else:
            print("Invalid command. Please try again.")

if __name__ == "__main__":
    main()
````
